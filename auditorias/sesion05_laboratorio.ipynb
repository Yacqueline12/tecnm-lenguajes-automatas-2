## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion05_c.c`. El archivo ya trae la
firma y un `TODO`: edita el contenido de la celda `%%writefile` que está abajo (o el

````python
%%writefile api_sesion05.h
/* api_sesion05.h - contrato de la Sesion 05
 * Recorrido postorden y RPN
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION05_H
#define API_SESION05_H

typedef enum { NODO_NUM, NODO_BINOP } TipoNodo;
typedef struct ASTNode {
    TipoNodo tipo;
    int valor;
    char op;
    struct ASTNode *izq;
    struct ASTNode *der;
} ASTNode;

void postorden(ASTNode *n, char *buf, int cap);

#endif /* API_SESION05_H */
````

````python
%%writefile artesano_sesion05_c.c
#include "api_sesion05.h"
#include <stdio.h>

static void escribir(ASTNode *n, char *buf, int cap, int *pos)
{
    char dato[32];
    int i = 0;

    if (n == NULL)
        return;

    escribir(n->izq, buf, cap, pos);
    escribir(n->der, buf, cap, pos);

    if (*pos >= cap - 1)
        return;

    if (n->tipo == NODO_NUM)
        snprintf(dato, sizeof(dato), "%d", n->valor);
    else
        snprintf(dato, sizeof(dato), "%c", n->op);

    if (*pos != 0) {
        if (*pos < cap - 1)
            buf[(*pos)++] = ' ';
        else
            return;
    }

    while (dato[i] != '\0' && *pos < cap - 1) {
        buf[*pos] = dato[i];
        (*pos)++;
        i++;
    }

    buf[*pos] = '\0';
}

void postorden(ASTNode *n, char *buf, int cap)
{
    int pos = 0;

    if (buf == NULL || cap <= 0)
        return;

    buf[0] = '\0';
    escribir(n, buf, cap, &pos);
}

````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion05_c.c`. La

 Contexto: estoy trabajando en la sesión 05 de Lenguajes y Autómatas II. La actividad consiste en recorrer un árbol AST para obtener una expresión en notación polaca inversa. Estoy aprendiendo a programar en C y necesito que el código sea fácil de entender.

 Rol: quiero que actúes como un programador que me ayude a resolver el ejercicio y que me explique las instrucciones de una manera sencilla, sin utilizar cosas demasiado complicadas.

 Especificación: necesito crear la función postorden en C para recorrer el árbol primero por la izquierda, luego por la derecha y al final la raíz. El resultado debe guardarse en el buffer con espacios entre los números y operadores, respetando su capacidad, funcionando con un árbol NULL y sin utilizar printf ni main().

 Output: dame el código completo para pegarlo en ia_sesion05_c.c. Utiliza una solución diferente a la que hice manualmente, pero que cumpla con las mismas condiciones. Explícame brevemente cómo funciona y revisa que pueda compilarse con -std=c11 -Wall -Wextra -Werror.

````python
%%writefile ia_sesion05_c.c
#include "api_sesion05.h"
#include <stdio.h>
#include <string.h>

static void agregar_nodo(ASTNode *n, char *buf, int cap, int *pos) {
    char token[32];
    int escritos;
    size_t largo;
    size_t espacio;

    if (n == NULL || *pos >= cap - 1) {
        return;
    }

    agregar_nodo(n->izq, buf, cap, pos);
    agregar_nodo(n->der, buf, cap, pos);

    if (*pos >= cap - 1) {
        return;
    }

    if (n->tipo == NODO_NUM) {
        escritos = snprintf(token, sizeof(token), "%d", n->valor);
    } else {
        escritos = snprintf(token, sizeof(token), "%c", n->op);
    }

    if (escritos < 0 || (size_t)escritos >= sizeof(token)) {
        return;
    }

    largo = strlen(token);
    espacio = (*pos > 0) ? 1u : 0u;

    /* Comprobar que el token completo cabe en el buffer */
    if ((size_t)*pos + espacio + largo >= (size_t)cap) {
        *pos = cap - 1;
        buf[*pos] = '\0';
        return;
    }

    if (espacio > 0) {
        buf[(*pos)++] = ' ';
    }

    memcpy(buf + *pos, token, largo + 1);
    *pos += (int)largo;
}

void postorden(ASTNode *n, char *buf, int cap) {
    int pos = 0;

    if (buf == NULL || cap <= 0) {
        return;
    }

    buf[0] = '\0';
    agregar_nodo(n, buf, cap, &pos);
}
````

````python
%%writefile harness_sesion05_c.c
/* harness_sesion05_c.c - auditoria automatica de la Sesion 05
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion05_c.c artesano_sesion05_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion05.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 05 - Recorrido postorden y RPN ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { ASTNode a = {NODO_NUM, 42, 0, NULL, NULL}; char b[64]; postorden(&a, b, 64); verificar_str("hoja 42", "42", b); }
    { ASTNode a = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 5, 0, NULL, NULL}; ASTNode c = {NODO_NUM, 2, 0, NULL, NULL}; ASTNode m = {NODO_BINOP, 0, '*', &b, &c}; ASTNode s = {NODO_BINOP, 0, '+', &a, &m}; char buf[64]; postorden(&s, buf, 64); verificar_str("3+(5*2)", "3 5 2 * +", buf); }
    { char b[16]; postorden(NULL, b, 16); verificar_str("arbol NULL", "", b); }
    { ASTNode a = {NODO_NUM, 1, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 2, 0, NULL, NULL}; ASTNode c = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode d = {NODO_NUM, 4, 0, NULL, NULL}; ASTNode s1 = {NODO_BINOP, 0, '+', &a, &b}; ASTNode s2 = {NODO_BINOP, 0, '+', &c, &d}; ASTNode m = {NODO_BINOP, 0, '*', &s1, &s2}; char buf[64]; postorden(&m, buf, 64); verificar_str_largo("(1+2)*(3+4)", "1 2 + 3 4 + *", buf); }
    { ASTNode a = {NODO_NUM, 8, 0, NULL, NULL}; ASTNode b = {NODO_NUM, 3, 0, NULL, NULL}; ASTNode r = {NODO_BINOP, 0, '-', &a, &b}; char buf[32]; postorden(&r, buf, 32); verificar_str("8-3", "8 3 -", buf); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion05.md" "\n");
    return 0;
}
````
