````python
%%writefile artesano_sesion26_c.c
/* artesano_sesion26_c.c - Fase 1: El Artesano (sin IA)
 * Llamadas, parámetros y retorno.
 *
 * Implementa lo que declara api_sesion26.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion26.h"
#include <stdio.h>
#include <string.h>

/* Agrega un texto al final de una cadena */
int agregar_texto(char *destino, int limite, int *posicion,
                  const char *texto) {

    int i = 0;

    if (texto == NULL) {
        return 0;
    }

    while (texto[i] != '\0') {

        if (*posicion >= limite - 1) {
            return 0;
        }

        destino[*posicion] = texto[i];
        (*posicion)++;
        i++;
    }

    destino[*posicion] = '\0';

    return 1;
}

/* Construye manualmente el numero de argumentos */
int agregar_numero(char *destino, int limite, int *posicion,
                   int numero) {

    if (numero < 0 || numero > 9) {
        return 0;
    }

    if (*posicion >= limite - 1) {
        return 0;
    }

    destino[*posicion] = (char)('0' + numero);
    (*posicion)++;

    destino[*posicion] = '\0';

    return 1;
}

/* Genera PARAM y CALL */
int emitir_llamada(char lineas[][64], int max, int *n,
                   const char *funcion, const char *a1,
                   const char *a2, const char *resultado) {

    char nuevas[3][64] = {{0}};
    int cantidad = 0;
    int posicion = 0;
    int i;

    if (lineas == NULL || n == NULL ||
        funcion == NULL || max <= 0 ||
        *n < 0 || *n > max) {
        return 0;
    }

    /* Primero guardamos los argumentos */
    if (a1 != NULL) {

        if (!agregar_texto(nuevas[cantidad], 64,
                           &posicion, "PARAM ") ||
            !agregar_texto(nuevas[cantidad], 64,
                           &posicion, a1)) {
            return 0;
        }

        cantidad++;
        posicion = 0;
    }

    if (a2 != NULL) {

        if (!agregar_texto(nuevas[cantidad], 64,
                           &posicion, "PARAM ") ||
            !agregar_texto(nuevas[cantidad], 64,
                           &posicion, a2)) {
            return 0;
        }

        cantidad++;
        posicion = 0;
    }

    /* Preparamos la instruccion CALL */
    if (resultado != NULL) {

        if (!agregar_texto(nuevas[cantidad], 64,
                           &posicion, resultado) ||
            !agregar_texto(nuevas[cantidad], 64,
                           &posicion, " = ")) {
            return 0;
        }
    }

    if (!agregar_texto(nuevas[cantidad], 64,
                       &posicion, "CALL ") ||
        !agregar_texto(nuevas[cantidad], 64,
                       &posicion, funcion) ||
        !agregar_texto(nuevas[cantidad], 64,
                       &posicion, ", ") ||
        !agregar_numero(nuevas[cantidad], 64,
                        &posicion, cantidad)) {
        return 0;
    }

    cantidad++;

    /* Revisamos si hay espacio suficiente */
    if (cantidad > max - *n) {
        return 0;
    }

    /* Copiamos las instrucciones al arreglo original */
    for (i = 0; i < cantidad; i++) {
        strcpy(lineas[*n + i], nuevas[i]);
    }

    *n += cantidad;

    return cantidad;
}

/* Genera una instruccion RETURN */
void emitir_return(char *dst, int cap, const char *valor) {

    int posicion = 0;

    if (dst == NULL || cap <= 0) {
        return;
    }

    dst[0] = '\0';

    if (!agregar_texto(dst, cap, &posicion, "RETURN")) {
        dst[0] = '\0';
        return;
    }

    if (valor != NULL) {

        if (!agregar_texto(dst, cap, &posicion, " ") ||
            !agregar_texto(dst, cap, &posicion, valor)) {
            dst[0] = '\0';
            return;
        }
    }
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle

Contexto: estoy trabajando en Lenguajes y Autómatas II, utilizando el lenguaje C99 con GCC. Necesito desarrollar funciones que permitan representar llamadas a funciones y retornos mediante código de tres direcciones (TAC).

Especificación: desarrolla las funciones `emitir_llamada()` y `emitir_return()` respetando exactamente las firmas declaradas en `api_sesion26.h`.

La función `emitir_llamada()` debe:
- Generar una instrucción PARAM por cada argumento diferente de NULL.
- Mantener el orden de los argumentos.
- Generar CALL indicando la cantidad real de parámetros.
- Permitir llamadas sin argumentos.
- Permitir llamadas sin variable de resultado.
- Conservar correctamente el contador de instrucciones para soportar llamadas anidadas.
- Validar los límites del arreglo antes de escribir.

La función `emitir_return()` debe:
- Generar RETURN acompañado de un valor cuando exista.
- Generar únicamente RETURN cuando el valor sea NULL.
- Evitar escribir fuera de la capacidad del destino.

Output: entrega código modular en C99, sin función `main()`, compatible con `-Wall -Wextra -Werror`. Incluye una explicación de la lógica y un análisis de complejidad temporal y espacial.

````python
%%writefile ia_sesion26_c.c
/* ia_sesion26_c.c - Fase 2: El Cientifico
 *
 * Pega aqui el codigo que te devolvio la IA, con la MISMA firma
 * que declara api_sesion26.h. No puede traer main().
 */
#include "api_sesion26.h"
#include <stdio.h>
#include <string.h>

int emitir_llamada(char lineas[][64], int max, int *n,
                   const char *funcion, const char *a1,
                   const char *a2, const char *resultado) {

    char instrucciones[3][64] = {{0}};
    int parametros = 0;
    int total = 0;
    int longitud;

    if (lineas == NULL || n == NULL || funcion == NULL ||
        max <= 0 || *n < 0 || *n > max) {
        return 0;
    }

    /* Se generan los parametros en el orden recibido */
    if (a1 != NULL) {
        longitud = snprintf(instrucciones[total], 64,
                            "PARAM %s", a1);

        if (longitud < 0 || longitud >= 64) {
            return 0;
        }

        total++;
        parametros++;
    }

    if (a2 != NULL) {
        longitud = snprintf(instrucciones[total], 64,
                            "PARAM %s", a2);

        if (longitud < 0 || longitud >= 64) {
            return 0;
        }

        total++;
        parametros++;
    }

    /* Se construye la instruccion de llamada */
    if (resultado != NULL) {
        longitud = snprintf(instrucciones[total], 64,
                            "%s = CALL %s, %d",
                            resultado, funcion, parametros);
    } else {
        longitud = snprintf(instrucciones[total], 64,
                            "CALL %s, %d",
                            funcion, parametros);
    }

    if (longitud < 0 || longitud >= 64) {
        return 0;
    }

    total++;

    /* Se comprueba la capacidad disponible */
    if (total > max - *n) {
        return 0;
    }

    /* Se transfieren las instrucciones al arreglo final */
    memcpy(&lineas[*n], instrucciones,
           (size_t)total * sizeof instrucciones[0]);

    *n += total;

    return total;
}

void emitir_return(char *dst, int cap, const char *valor) {

    int longitud;

    if (dst == NULL || cap <= 0) {
        return;
    }

    if (valor != NULL) {
        longitud = snprintf(dst, (size_t)cap,
                            "RETURN %s", valor);
    } else {
        longitud = snprintf(dst, (size_t)cap,
                            "RETURN");
    }

    if (longitud < 0 || longitud >= cap) {
        dst[0] = '\0';
    }
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,

````python
%%writefile harness_sesion26_c.c
/* harness_sesion26_c.c - auditoria automatica de la Sesion 26
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion26_c.c artesano_sesion26_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion26.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 26 - Llamadas, parametros y retorno ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str("PARAM x", "PARAM x", l[0]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str("PARAM y", "PARAM y", l[1]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str_largo("CALL con dos argumentos", "_t0 = CALL f, 2", l[2]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "g", "x", NULL, "_t0"); emitir_llamada(l, 16, &n, "f", "_t0", NULL, "_t1"); verificar_str_largo("llamada anidada f(g(x))", "_t1 = CALL f, 1", l[3]); }
    { char l[8][64] = {{0}}; int n = 0; emitir_llamada(l, 8, &n, "p", NULL, NULL, "_t0"); verificar_str_largo("CALL sin argumentos", "_t0 = CALL p, 0", l[0]); }
    { char b[32] = ""; emitir_return(b, 32, "_t0"); verificar_str("RETURN con valor", "RETURN _t0", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion26.md" "\n");
    return 0;
}
````
